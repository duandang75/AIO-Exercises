```python
def color2gray_lightness(pixel_vector):
  return np.max(pixel_vector)*0.5 + np.min(pixel_vector)*0.5

img = mpimg.imread('/content/dog.jpeg')

gray_img_01 = np.apply_along_axis(color2gray_lightness, axis=2, arr=img)
print(gray_img_01[0,0])
```

```python
# Question 13
```

```python
def color2gray_avg(pixel_vector):
  return np.sum(pixel_vector) / 3

gray_img_02 = np.apply_along_axis(color2gray_avg, axis=2, arr=img)

print(gray_img_02[0,0])
```

```python
# Question 14
```

```python
def color2gray_lumi(pixel_vector):
  return np.sum(pixel_vector * np.array([0.21, 0.72, 0.07]))

gray_img_03 = np.apply_along_axis(color2gray_lumi, axis=2, arr=img)

print(gray_img_03[0,0])
```

```python
df = pd.read_csv('/content/advertising.csv')
data = df.to_numpy()
sales_data = data[:,3]
max_sales = sales_data.max()
max_idx = np.argmax(sales_data)
print(max_idx,max_sales)
```

```python
# Question 16
```

```python
mean_tv = np.mean(data[:,0])
print(mean_tv)
```

```python
# Question 17
```

```python
sales_cond = sales_data >= 20
sales_gt_20_count = np.count_nonzero(sales_data[sales_cond])
print(sales_gt_20_count)
```

```python
# Question 18
```

```python
sales_gt_15 = sales_data >= 15
radio_data = data[:,1]
radio_gt_15 = radio_data[sales_gt_15]
radio_gt_15_mean = np.mean(radio_gt_15)
print(radio_gt_15_mean)
```

```python
# Question 19
```

```python
news_data = data[:,2]
news_cond = news_data > news_data.mean()
sales_total = np.sum(sales_data[news_cond])
print(sales_total)
```

```python
# Question 20
```

```python
mean_sales = sales_data.mean()
scores = np.where(sales_data < mean_sales, "Bad",np.where(sales_data>mean_sales,"Good","Average"))
print(scores[7:10])
```

```python
# Question 21
```

```python
sales_mean = sales_data.mean()
sub_mean = abs(sales_data - sales_mean)
avg_idx = np.argmin(sub_mean)
sales_avg = sales_data[avg_idx]
score_sales = np.where(sales_data < sales_avg, "Bad",np.where(sales_data>sales_avg,"Good","Average"))
print(score_sales[7:10])
```
